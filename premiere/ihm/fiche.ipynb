## HTML : la structure d'une page

HTML sert à créer la structure et le contenu d'une page web.

Structure de base :

```html
<!DOCTYPE html>                 // Indique que le fichier est en HTML.
<html>
    <head>                      // Ouvre la balise des métadonnées.

        <title>Ma page</title>

    </head>
    <body> // Ouvre la balise du corps du site (les éléments affichables).

        <h1>Mon titre</h1>
        <p>Mon paragraphe.</p>

    </body>
</html>
```

## HTML : les balises
Les balises sont des éléments qui représentent un conteneur dont les attributs peuvent être modifiés. La balise générique est appelée :

```html
<div></div>
```
Elle représente un conteneur vide que l'on peut modifier comme bon nous semble.
::: {.callout-warning}
    ## Attention
    Les balises doivent toutes être fermées de cette manière :
    ```
    <nom_balise> </nom_balise>
    ``` de manière à toujours fermer le conteneur. Il existe des exceptions comme les images.
:::

Pour plus de facilité, il existe des balises déjà prêtes comme :

|Balise	|Utilité|
|-----|-----|
|```<h1> à <h6>```|Titres|
|```<p>```|Paragraphe|
|```<a>```|Lien|
|```<img>```|Image|
|```<ul>```|Liste à puces|
|```<ol>```|Liste numérotée|
|```<li>```|Élément d'une liste|
|```<strong>```|Texte important / gras|
|```<em>```|Texte mis en valeur / italique|
|```<div>```|Bloc générique|
|```<table>```|Grille d'éléments|

### Syntaxe
Les liens
```html
<a href="https://example.com">Clique ici</a>
```
* href indique la destination du lien.
* tout ce qui est dans la balise, si l'on clique dessus, nous redirigera.

Les images
```html
<img src="image.jpg" alt="Description de l'image">
```
* src → chemin de l'image
* alt → description de l'image

Les listes
```html
<ul>
    <li> élément 1 </li>
    <li> élément 2 </li>
</ul>
```
* 2 types de liste

Les tables
```html
<table>
    <tr>                          // table row 
        <th>titre colonne 1</th>   // table header
        <th>titre colonne 2</th>
    </tr>
    <tr>
        <td>élément colonne 1</td>  // table data
        <td>élément colonne 2</td>
    </tr>
</table>
```
## CSS : mettre en forme

CSS sert à modifier l'apparence d'une page HTML : couleurs, tailles, positions, etc.
Relier un fichier CSS

Dans le <head> (métadonnées) :
```html
<link rel="stylesheet" href="style.css">
```

Puis dans style.css :
```css
p {
    color: blue;
    font-size: 20px;
}
```
Syntaxe :
```css
sélecteur {
    propriété: valeur;
}
```

### Les classes et les IDs

Classe :

HTML
```html
<p class="important">Bonjour</p>
```
CSS
```css
.important {
    color: red;
}
```

ID :

HTML
```html
<p id="important">Bonjour</p>
```
CSS
```css
#important {
    color: red;
}
```

## JavaScript (pour aller plus loin)

JavaScript sert à ajouter de l'interactivité au site.
Relier un fichier JavaScript

Dans le <head> (métadonnées) :
```html
<script src="script.js"></script>
```
Exemple :
```js
window.addEventListener('DOMContentLoaded', () => {
  
});
```
* le fichier JS et le HTML sont séparés et peuvent être chargés l'un avant l'autre ; la ligne ci-dessus exécute le code après que la page soit chargée.

```js
let valeur = 0;
const compteur_like = document.getElementById("compteur_like");
const bouton_like = document.getElementById("incrementer_like");
```
* la fonction document.getElementById() va trouver la balise avec l'id correspondant et l'assigner à la variable.

```js
bouton_like.addEventListener("click", () => {
    //code
});
```
* on ajoute à la variable bouton_like (la balise avec l'id incrementer_like) un événement à chaque clic qui exécutera le code.

```js
valeur++;
compteur_like.textContent = valeur;
```
* à chaque clic, on ajoute donc 1 à valeur et on écrit dans compteur_like (la balise avec l'id compteur_like) la valeur.
